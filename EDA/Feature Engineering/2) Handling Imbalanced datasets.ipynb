#### Balancing imbalanced dataset

dataset -> 900 yes and 100 no -> the model is gonna have bias towards 900 yes hence we balancce the data set 
using 

1. up sampling : increase data points of the min value
2. down sampling : increase the data points of the max value

```python
import numpy as np
import pandas as pd


  # set the random seed for reproducibility
np.random.seed(123)
n_samples=1000
class_0_ratio=0.9
n_class_0=int(n_samples* class_0_ratio)
n_class_1= n_samples-n_class_0

```

```python
n_class_0, n_class_1
```

```python
## creating a dataframe with imbalance data set 


class_0= pd.DataFrame({
    'feature_1': np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2': np.random.normal(loc=0,scale=1,size=n_class_0),
    'target':[0]*n_class_0
})
class_1=pd.DataFrame({
    'feature_1': np.random.normal(loc=0,scale=1,size=n_class_1),
    'feature_2': np.random.normal(loc=0,scale=1,size=n_class_1),
    'target':[1]*n_class_1
})



```

```python
df=pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df
```

```python
df.head()
```

```python
df['target'].value_counts()
```

```python
## upsampling 
df_minority=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
from sklearn.utils import resample
# extrapolates the minority 

df_upsampled=resample(df_minority,replace=True,n_samples=len(df_majority),random_state=42)
df_upsampled.shape
```

```python
df_upsampled.head()
```

```python
df_up_sampled=pd.concat([df_majority,df_upsampled]).reset_index(drop=True)
```
