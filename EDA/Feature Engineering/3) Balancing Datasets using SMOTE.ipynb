#### SMOTE ( Synthetic Minority Oversampling Technique ) 


- used when minority has significantly fewer instances than majority class 
- works by generating synthetic instances of the minority class by interpolating between existing instances
- 

```python
from sklearn.datasets import make_classification
import numpy as np
import pandas as pd
```

```python
x,y=make_classification(n_samples=1000,n_features=2,n_clusters_per_class=1,weights=[0.90],random_state=12,n_redundant=0)
```

```python
import pandas as pd
df1=pd.DataFrame(x,columns=['f1','f2'])
df2=pd.DataFrame(y,columns=['target'])
final_df=pd.concat([df1,df2],axis=1)
final_df['target'].value_counts()
```

```python
import matplotlib.pyplot as plt
plt.scatter(final_df['f1'],final_df['f2'],c=final_df['target'])
```

```python
from imblearn.over_sampling import SMOTE
oversample=SMOTE()
```

```python
## Transform the data set 
x,y=oversample.fit_resample(
    final_df[['f1', 'f2']],
    final_df['target']
)

x.head()
```

```python
df1=pd.DataFrame(x,columns=['f1','f2'])
df2=pd.DataFrame(y,columns=['target'])
oversample_df=pd.concat([df1,df2],axis=1)
oversample_df.head()
```
