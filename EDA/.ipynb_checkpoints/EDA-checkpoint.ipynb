#### Feature Scaling
##### When we are dealing with a dataset containing features in terms of range of values / units 
##### then we have to normalize / standardize the data such that the data doesn't effect the outcome
##### so that the difference in the scale doesn't effect the outcome
##### predictive modeling predictive analysis

#### Normalization is a data preprocessing technique that rescales data into a fixed range, usually: 0 and 1
#### Standardization is a data preprocessing technique that transforms data so that it has:

#### Mean = 0
#### Standard deviation = 1


#### Formula  
##### Standard scaler (digestion): Z= ((x-u(mean)) / Standard Deviation)
##### Min-Max Scaler(Normalisation) :  X  = (x-Xmin)/(Xmax-Xmin)
#### for standardised data mean will be almost zero
#### standardization is done based on standard normal distribution

# Outlier Treatment
#### Outliers are the extreme values in the data its is an abnormal observation that deviate from the nom and does not fit in the normal behavior of the data
##### Detect outliers using : BOX plot, Histogram, Scatter Plot, Z-score, Interquartile range (values outs of 1.5 time of IQR)
##### we handle outlier using following methods 
###### Remove the outliers
###### Replace outlier with suitable values by using following methods: quartile method, inter quartile method
###### use that ML method which is not sensitive to outliers
###### Like KNN, Decision Tree, SVM, Naive Bayes, Ensemble methods
#### how to handle

##### Time series forcasting ( eg: temp of country noting everyday some of the temps could be outliers we should not drop the outliers each data is important ) 
##### When we are dealing with data set where 1 or 2 people as is outlier in that case we can remove or delete them ( check each case ) 



#### 3 sigma techniques ( Outlier treatment )  ( 3 Sd technique ) 
#### Bases on principles of normal distribution 
#### take is the data lies beyond the range of -3 SD to +3 SD then there are outliers 


```python
import statistics
```

```python
df=pd.read_csv(r"C:\Users\DataScience\Desktop\DATA_SCIENCE\EDA\EDA-main\Data\raw_sales.csv")
df.info()
```

```python
df.head()
```

```python
def find_anomalies(data):
    anomalies=[]
    random_data_std=statistics.stdev(data)
    random_data_mean=statistics.mean(data)
    
    anamoly_cut_off=random_data_std*3
    lower_limit=random_data_std-anamoly_cut_off
    upper_limit=random_data_std+anamoly_cut_off

    for outlier in data:
        if outlier< lower_limit or outlier > upper_limit:
            anomalies.append(outlier)
    print(anomalies)

    
```

```python
find_anomalies(df['price'])
```
